```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("StudentsPerformance.csv")


# Create 2 plots
fig, axes = plt.subplots(1, 2, figsize=(14, 6))


# 1. Math Score Violin Plot
sns.violinplot(
    data=df,
    x="gender",
    y="math score",
    hue="test preparation course",
    split=True,
    ax=axes[0]
)

axes[0].set_title("Math Score by Gender and Test Preparation")
axes[0].set_xlabel("Gender")
axes[0].set_ylabel("Math Score")


# 2. Reading Score Violin Plot
sns.violinplot(
    data=df,
    x="gender",
    y="reading score",
    hue="test preparation course",
    split=True,
    ax=axes[1]
)

axes[1].set_title("Reading Score by Gender and Test Preparation")
axes[1].set_xlabel("Gender")
axes[1].set_ylabel("Reading Score")


# Main title
fig.suptitle(
    "Student Exam Scores by Gender and Test Preparation"
)

plt.tight_layout()
plt.show()
```
